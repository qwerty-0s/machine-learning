```python
import numpy as np
import matplotlib.pyplot as plt
import math
import pandas as pd

df = pd.DataFrame()
df = pd.read_csv('/content/weather_data.csv', sep=',')
length = len(df['Location'])
sumDD=0
sumDT=0
sumDH=0
sumDP=0
sumDW=0
sumDB=0
sum=0
sumprev=0
D=0.1 # дата, температура, влажность, осадки, ветер, отступ
T=0
H=0
P=0
W=0
B=0
step=0.001

def Ds(str):
    n=str.split()
    date=n[0].split("-")
    time=n[1].split(":")
    sum=int(date[1])*30*24*60*60+int(date[2])*24*60*60+int(time[0])*60*60+int(time[1])*60+int(time[2]) #дату в секунды
    return sum


def func(D,di,T,ti,H,hi,P,pi,W,wi,B):
    return (D*di+T*ti+H*hi+P*pi+W*wi+B)**2/(D**2+T**2+H**2+P**2+W**2) #частная производная от функции расстояния от точки до плоскости
def funcDD(D,di,T,ti,H,hi,P,pi,W,wi,B):
    return (2*(D*di+T*ti+H*hi+P*pi+W*wi+B)*di*(D**2+T**2+H**2+P**2+W**2)-(D*di+T*ti+H*hi+P*pi+W*wi+B)**2*2*D)/((D**2+T**2+H**2+P**2+W**2)**2)
def funcDT(D,di,T,ti,H,hi,P,pi,W,wi,B):
    return (2*(D*di+T*ti+H*hi+P*pi+W*wi+B)*ti*(D**2+T**2+H**2+P**2+W**2)-(D*di+T*ti+H*hi+P*pi+W*wi+B)**2*2*T)/((D**2+T**2+H**2+P**2+W**2)**2)
def funcDH(D,di,T,ti,H,hi,P,pi,W,wi,B):
    return (2*(D*di+T*ti+H*hi+P*pi+W*wi+B)*hi*(D**2+T**2+H**2+P**2+W**2)-(D*di+T*ti+H*hi+P*pi+W*wi+B)**2*2*H)/((D**2+T**2+H**2+P**2+W**2)**2)
def funcDP(D,di,T,ti,H,hi,P,pi,W,wi,B):
    return (2*(D*di+T*ti+H*hi+P*pi+W*wi+B)*pi*(D**2+T**2+H**2+P**2+W**2)-(D*di+T*ti+H*hi+P*pi+W*wi+B)**2*2*P)/((D**2+T**2+H**2+P**2+W**2)**2)
def funcDW(D,di,T,ti,H,hi,P,pi,W,wi,B):
    return (2*(D*di+T*ti+H*hi+P*pi+W*wi+B)*wi*(D**2+T**2+H**2+P**2+W**2)-(D*di+T*ti+H*hi+P*pi+W*wi+B)**2*2*W)/((D**2+T**2+H**2+P**2+W**2)**2)
def funcDB(D,di,T,ti,H,hi,P,pi,W,wi,B):
    return (2*(D*di+T*ti+H*hi+P*pi+W*wi+B)*1*(D**2+T**2+H**2+P**2+W**2))/((D**2+T**2+H**2+P**2+W**2)**2)

NewYork = pd.DataFrame({
     'D': [],
     'T': [],
     'H': [],
     'P': [],
     'W' :[],
})

s=0
for i in range(10000):
    if df.loc[i, 'Location']=='New York':
        NewYork.loc[s,'D']=Ds(df.loc[i, 'Date_Time'])
        NewYork.loc[s,'T']=df.loc[i, 'Temperature_C']
        NewYork.loc[s,'H']=df.loc[i, 'Humidity_pct']
        NewYork.loc[s,'P']=df.loc[i, 'Precipitation_mm']
        NewYork.loc[s,'W']=df.loc[i, 'Wind_Speed_kmh']
        s=s+1


i=0
N=len(NewYork['D'])
print(N)

ny=np.zeros((N,5),float)
ny=NewYork.to_numpy()

print(ny)

while 1:  #градиентный спуск для поиска коэффициентов

    sumDD=0
    sumDT=0
    sumDH=0
    sumDP=0
    sumDW=0
    sumDB=0
    sum=0
    i=0

    for i in range(20):   #должно быть N но считате очень долго
        sumDD=sumDD+funcDD(D,ny[i][0],T,ny[i][1],H,ny[i][2],P,ny[i][3],W,ny[i][4],B)
        sumDT=sumDT+funcDT(D,ny[i][0],T,ny[i][1],H,ny[i][2],P,ny[i][3],W,ny[i][4],B)
        sumDH=sumDH+funcDH(D,ny[i][0],T,ny[i][1],H,ny[i][2],P,ny[i][3],W,ny[i][4],B)
        sumDP=sumDP+funcDP(D,ny[i][0],T,ny[i][1],H,ny[i][2],P,ny[i][3],W,ny[i][4],B)
        sumDW=sumDW+funcDW(D,ny[i][0],T,ny[i][1],H,ny[i][2],P,ny[i][3],W,ny[i][4],B)
        sumDB=sumDB+funcDB(D,ny[i][0],T,ny[i][1],H,ny[i][2],P,ny[i][3],W,ny[i][4],B)
        sum=sum+func(D,ny[i][0],T,ny[i][1],H,ny[i][2],P,ny[i][3],W,ny[i][4],B)

    if sumprev-sum <= 0.01 and sumprev != 0:
        break

    D=D-sumDD*step
    T=T-sumDT*step
    H=H-sumDH*step
    P=P-sumDP*step
    W=W-sumDW*step
    B=B-sumDB*step
    sumprev=sum


print(D,T,H,P,W,B)

print((-4820674*D-(-6.86)*T-30*H-0*W-B)/P)  # тест
```
